```python
print("Number 6 \n Вектор q(a, 2) має модуль 5. Знайти всі значення a.|q| = sqrt(a^2 + 2^2) = 5  =>  a^2 = 25 - 4 = 21  =>  a = ±sqrt(21)\n")
modulus = 5
known_component = 2
a_squared = modulus**2 - known_component**2
a_positive = math.sqrt(a_squared)
a_negative = -math.sqrt(a_squared)
print(f"Можливі значення a: {a_positive} та {a_negative}")
```
