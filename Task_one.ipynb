```python
import matplotlib.pyplot as plt
```

```python


months = ["Січ", "Лют", "Бер", "Кві", "Тра", "Чер", "Лип", "Сер", "Вер", "Жов", "Лис", "Гру"]
plan = [100, 120, 130, 110, 140, 150, 160, 155, 145, 170, 180, 200]
fakt = [95, 115, 125, 120, 135, 160, 150, 165, 140, 175, 190, 195]

plt.plot(months, plan, label="План", marker="o")
plt.plot(months, fakt, label="Факт", marker="o")

plt.xlabel("Місяць")
plt.ylabel("Продажі")
plt.title("Порівняння планових та фактичних продажів")
plt.legend()
plt.show()
```
