```python
import pandas as pd
```

```python
def Books():
    book_data = {
    'Title': ['Romeo and Juliet','The Mysterious Affair at Styles', "Harry Potter and the Philosopher's Stone", 'One Piece','Bleach','Hellsing','The Hunger Games'],
    'Author': ['William Shakespeare','Agatha Christie', 'J. K. Rowling', 'Eiichiro Oda','Tite Kubo','Kouta Hirano','Suzanne Collins'],
    'Year of publishing':[1597,1920,1997,1997,2001,1997,2008],
    'Price':[100,100,120,20,50,50,70]
    }
    table = pd.DataFrame(book_data)
    print("\n-------------------NOT SORTED Table-------------------\n")
    print(table)
    
    books_after_2015 = table[table['Year of publishing'] > 2015]
    print("\n-------------------SORTED BY YEAR Table-------------------\n")
    print(books_after_2015)
    
    print("\n-------------------SORTED BY PRICE Table-------------------\n")
    print(table.sort_values(by=["Price"]))
Books()
```
