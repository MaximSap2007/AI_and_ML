```python
import matplotlib.pyplot as plt
```

```python

dates = ["2024-06-01", "2024-06-02", "2024-06-03", "2024-06-04", "2024-06-05", "2024-06-06", "2024-06-07"]
temperature = [22, 24, 23, 25, 27, 26, 24]
humidity = [60, 58, 65, 55, 50, 52, 57]

plt.figure(figsize=(10, 5))
plt.plot(dates, temperature, marker="o", label="Температура, °C")
plt.plot(dates, humidity, marker="s", label="Вологість, %")
plt.xlabel("Дата")
plt.ylabel("Значення")
plt.title("Температура та вологість за тиждень")
plt.legend()
plt.xticks(rotation=45)
plt.show()
```
