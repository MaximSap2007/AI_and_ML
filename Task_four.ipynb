```python
import pandas as pd
```

```python
employees = pd.DataFrame({
    'Імя': ['Іван', 'Олена', 'Сергій', 'Марія', 'Іван', 'Олена', 'Андрій', 'Сергій'],
    'Проект': ['Альфа', 'Бета', 'Альфа', 'Гамма', 'Бета', 'Альфа', 'Гамма', 'Бета'],
    'Години': [10, 15, 8, 12, 6, 9, 14, 11]
})
print(employees)

# загальна кількість годин за кожним співробітником
hours_by_employee = employees.groupby('Імя')['Години'].sum()
print("Години за співробітником:\n", hours_by_employee)

# загальна кількість годин за кожним проектом
hours_by_project = employees.groupby('Проект')['Години'].sum()
print("Години за проектом:\n", hours_by_project)

# співробітник, що витратив найбільше годин
max_employee = hours_by_employee.idxmax()
print("Найбільше годин витратив:", max_employee, "-", hours_by_employee.max())
```
