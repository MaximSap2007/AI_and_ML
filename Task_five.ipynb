```python
import pandas as pd
```

```python
tickets = pd.DataFrame({
    'Фільм': ['Фільм А', 'Фільм Б', 'Фільм А', 'Фільм В', 'Фільм Б', 'Фільм А', 'Фільм В', 'Фільм Б', 'Фільм А', 'Фільм В', 'Фільм Б', 'Фільм А'],
    'Місто': ['Київ', 'Львів', 'Львів', 'Київ', 'Київ', 'Одеса', 'Львів', 'Одеса', 'Київ', 'Одеса', 'Київ', 'Львів'],
    'Продані квитки': [120, 95, 80, 150, 60, 110, 75, 90, 130, 85, 100, 70]
})
print(tickets)

# загальна кількість проданих квитків за кожним фільмом
tickets_by_movie = tickets.groupby('Фільм')['Продані квитки'].sum()
print("Квитки за фільмом:\n", tickets_by_movie)

# загальна кількість квитків за кожним містом
tickets_by_city = tickets.groupby('Місто')['Продані квитки'].sum()
print("Квитки за містом:\n", tickets_by_city)

# фільм з найбільшою кількістю продажів
best_movie = tickets_by_movie.idxmax()
print("Найбільше продажів у фільму:", best_movie, "-", tickets_by_movie.max())
```
