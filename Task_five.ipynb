```python
import matplotlib.pyplot as plt
```

```python


hours = list(range(24))
load = [10, 8, 6, 5, 5, 7, 15, 30, 45, 50, 55, 60, 62, 58, 55, 53, 50, 48, 40, 35, 28, 20, 15, 12]

plt.figure(figsize=(10, 5))
plt.plot(hours, load, color="darkorange")
plt.fill_between(hours, load, color="orange", alpha=0.3)
plt.grid(True)
plt.xlabel("Година")
plt.ylabel("Навантаження, %")
plt.title("Навантаження сервера за добу")
plt.show()
```
