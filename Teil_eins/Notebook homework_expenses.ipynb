```python
import pandas as pd
```

```python
"""
Вам необхідно розрахувати показники за списком щоденних витрат. Деталі реалізації:
Створіть Notebook homework_expenses.ipynb.
У кодовій комірці створіть список: [1200, 1500, 800, 600, 1000, 400, 900].
Розрахуйте:
кількість днів, коли витрата була вищою за 1000.
Відобразіть усі результати в таблиці

"""
array = [1200, 1500, 800, 600, 1000, 400, 900]
sum_array = sum(array)
min_array = min(array)
max_array = max(array)
middle = (array[0]+array[1]+array[2]+array[4]+array[5]+array[6]) / len(array)

count = 0
for i in array:
    if i > 1000:
        count +=1

tabel = pd.DataFrame({'Sum': [sum_array], 'Min' : [min_array], 'Max' : [max_array], 'Middle':[middle], 'The number of days when consumption was higher than 1000': [count]})
print(tabel)
        

    
```
