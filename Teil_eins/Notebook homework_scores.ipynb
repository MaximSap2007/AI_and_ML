```python
"""Порівняйте результати з предметів і обчисліть основні показники. Деталі реалізації:
Створіть Notebook homework_scores.ipynb.

У кодовій комірці створіть списки:
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87

Для кожного предмета обчисліть:

середній бал;

мінімальний бал;

максимальний бал.

Доробити нижню частину0
Додатково знайдіть предмет з найвищим середнім результатом (кодом, без тексту).

Представте результати в таблиці."""
import pandas as pd
```

```python
math = [80, 92, 75, 60, 88]
physics = [70, 65, 90, 85, 78]
english = [95, 100, 85, 90, 87]

data = {
    'Предмет': ['Math', 'Physics', 'English'],
    'Sum': [sum(math), sum(physics), sum(english)],
    'Max': [max(math), max(physics), max(english)],
    'Min': [min(math), min(physics), min(english)],
    'Avg': [sum(math)/len(math), sum(physics)/len(physics), sum(english)/len(english)]
}

table = pd.DataFrame(data)

print(table)

best_subject = table.loc[table['Avg'].idxmax(), 'Предмет']
print(f"\nНайкращий предмет за середнім балом: {best_subject}")
```
