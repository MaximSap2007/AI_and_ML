```python
csv = pd.read_csv("D:\ITSEP Python\AI_and_ML\orders.csv")
print(csv.head(10))
name = [0,0,0,0]

for i in range(len(csv)):
    if csv['Клієнт'][i] == "Іван Петренко":
        name[0]+=1
    elif csv['Клієнт'][i] == "Олена Коваль":
        name[1]+=1
    elif csv['Клієнт'][i] == "Сергій Бондар":
        name[2]+=1
    elif csv['Клієнт'][i] == "Марія Шевченко":
        name[3]+=1
users_bd = {
  'Іван Петренко':[name[0]],
  'Олена Коваль':[name[1]],
  'Сергій Бондар': [name[2]],
  'Марія Шевченко':[name[3]]
}
print(pd.DataFrame(users_bd))
print("Min:",min(csv['Сума']))
print("Max:",max(csv['Сума']))
print("Сума:",sum(csv['Сума']))
```
