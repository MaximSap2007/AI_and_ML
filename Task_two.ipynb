```python
import matplotlib.pyplot as plt
import numpy as np
```

```python


# 100 значень віку
np.random.seed(42)
ages = np.random.randint(18, 70, size=100)

mean_age = np.mean(ages)

plt.figure(figsize=(8, 5))
plt.hist(ages, bins=15, edgecolor="black")
plt.axvline(mean_age, color="red", linestyle="--", label=f"Середнє: {mean_age:.1f}")
plt.xlabel("Вік")
plt.ylabel("Кількість осіб")
plt.title("Розподіл вікових груп")
plt.legend()
plt.show()
```
