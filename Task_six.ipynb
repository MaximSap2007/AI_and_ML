```python
import matplotlib.pyplot as plt
```

```python


time_scale = list(range(1, 11))
conversion = [2.1, 2.3, 2.0, 2.5, 2.7, 2.6, 2.8, 3.0, 2.9, 3.1]
retention = [40, 42, 41, 45, 47, 46, 48, 50, 49, 51]
avg_check = [350, 360, 355, 370, 380, 375, 390, 400, 395, 410]
orders_count = [120, 125, 130, 128, 135, 140, 138, 145, 150, 155]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0, 0].plot(time_scale, conversion, marker="o", color="blue")
axes[0, 0].set_title("Конверсія, %")

axes[0, 1].plot(time_scale, retention, marker="o", color="green")
axes[0, 1].set_title("Утримання, %")

axes[1, 0].plot(time_scale, avg_check, marker="o", color="purple")
axes[1, 0].set_title("Середній чек")

axes[1, 1].plot(time_scale, orders_count, marker="o", color="brown")
axes[1, 1].set_title("Кількість замовлень")

fig.suptitle("Метрики продукту")
plt.tight_layout()
plt.show()
```
