```python
import pandas as pd
```

```python
customers = pd.DataFrame({
    'ID': [1, 2, 3, 4, 5],
    "Імя": ['Іван', 'Олена', 'Сергій', 'Марія', 'Андрій'],
    'Місто': ['Київ', 'Львів', 'Одеса', 'Харків', 'Київ']
})
print(customers)

orders = pd.DataFrame({
    'ID клієнта': [1, 2, 3, 1, 4, 5, 2],
    'Замовлення': ['Ноутбук', 'Телефон', 'Книга', 'Мишка', 'Стіл', 'Лампа', 'Навушники'],
    'Вартість': [25000, 15000, 300, 500, 3000, 800, 1200]
})
print(orders)

# обʼєднуємо таблиці за ідентифікатором клієнта
merged = orders.merge(customers, left_on='ID клієнта', right_on='ID')
print(merged)

# загальна вартість замовлень за кожним клієнтом
cost_by_customer = merged.groupby('Імя')['Вартість'].sum()
result = pd.DataFrame(cost_by_customer)
print(result)
```
