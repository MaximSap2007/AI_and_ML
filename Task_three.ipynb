```python
import pandas as pd
import numpy as np
```

```python
def Food():
    products = pd.DataFrame({
    'Продукт': ['Курка', 'Яловичина', 'Лосось', 'Гречка', 'Рис', 'Йогурт', 'Сир', 'Яблуко', 'Банан', 'Горіхи', 'Хліб', 'Яйця'],
    'Категорія': ['Мʼясо', 'Мʼясо', 'Риба', 'Крупи', 'Крупи', 'Молочні', 'Молочні', 'Фрукти', 'Фрукти', 'Горіхи', 'Випічка', 'Молочні'],
    'Калорії': [165, 250, 208, 343, 130, 59, 402, 52, 89, 607, 265, 155],
    'Білки': [31, 26, 20, 13, 2.7, 10, 25, 0.3, 1.1, 20, 9, 13]
    })
    print(products)
    weight = products[products['Калорії'] > 300]
    print("Усі продукти з калорійністю вище 300:\n",weight)
    
    type = np.float64([0,0,0,0,0,0,0])
    count = np.float64([0,0,0,0,0,0,0])
    for i in range(len(products)):
        if products['Категорія'][i] == "Мʼясо":
            type[0] += products['Білки'][i]
            count[0]+=1
        elif products['Категорія'][i] == "Риба":
            type[1] += products['Білки'][i]
            count[1]+=1
        elif products['Категорія'][i] == "Крупи":
            type[2] += products['Білки'][i]
            count[2]+=1
        elif products['Категорія'][i] == "Молочні":
            type[3] += products['Білки'][i]
            count[3]+=1
        elif products['Категорія'][i] == "Фрукти":
            type[4] += products['Білки'][i]
            count[4]+=1
        elif products['Категорія'][i] == "Горіхи":
            type[5] += products['Білки'][i]
            count[5]+=1
        elif products['Категорія'][i] == "Випічка":
            type[6] += products['Білки'][i]
            count[6]+=1
    avg = type / count
    type_bd = pd.DataFrame({
        'Мʼясо':[avg[0]],
        'Риба':[avg[1]],
        'Крупи':[avg[2]],
        'Молочні':[avg[3]],
        'Фрукти':[avg[4]],
        'Горіхи':[avg[5]], 
        'Випічка':[avg[6]],
    })
    print(type_bd)
Food()
    
```
