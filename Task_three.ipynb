```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
np.random.seed(1)
group_a = np.random.normal(75, 10, 30)
group_b = np.random.normal(68, 15, 30)
group_c = np.random.normal(82, 8, 30)

plt.figure(figsize=(8, 5))
plt.boxplot([group_a, group_b, group_c], tick_labels=["Група A", "Група B", "Група C"])
plt.xlabel("Група")
plt.ylabel("Бали")
plt.title("Розкид результатів іспитів за групами")
plt.show()
```
